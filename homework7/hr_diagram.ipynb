```python
# --- 3 Plot a Hertzsprung-Russell Diagram ---
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("star_data.txt", skiprows=1)
```

```python
print(np.shape(data))
# 2642 rows
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
# Minimums and maximums

print(np.min(magnitude))
row_min_magnitude = np.argmin(magnitude)
print(f"The minimum magnitude star is in row {row_min_magnitude}")

print(np.max(magnitude))
row_max_magnitude = np.argmax(magnitude)
print(f"The maximum magnitude star is in row {row_max_magnitude}")

print(np.min(temperature))
row_min_temperature = np.argmin(temperature)
print(f"The minimum temperature star is in row {row_min_temperature}")

print(np.max(temperature))
row_max_temperature = np.argmax(temperature)
print(f"The maximum temperature star is in row {row_max_temperature}")

print(f"The 1988th star in the list is the brightest and the 264th star in the list is the dimmest.")
```

```python
row_min_magnitude = np.where(magnitude == (np.min(magnitude)))
```

```python
# 3.3 Plot and HR Diagram

log_magnitude = np.log(magnitude)
log_temperature= np.log(temperature)

plt.figure(figsize = (8, 6))
plt.scatter(log_temperature, log_magnitude)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.show
```

```python
# Plot all points
plt.plot(figsize = (8, 6))
plt.scatter(log_temperature, log_magnitude)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

# Plot brightest star
min_magnitude_star_x = np.log(data[1988, 1])
min_magnitude_star_y = np.log(data[1988, 0])
plt.scatter(min_magnitude_star_x, min_magnitude_star_y, color = "black")

# Plot dimmest star
max_magnitude_star_x = np.log(data[264, 1])
max_magnitude_star_y = np.log(data[264, 0])
plt.scatter(max_magnitude_star_x, max_magnitude_star_y, color = "purple")

# Plot hottest star
max_temp_star_x = np.log(data[2613, 1])
max_temp_star_y = np.log(data[2613, 0])
plt.scatter(max_temp_star_x, max_temp_star_y, color = "green")

# Plot coldest star
min_temp_star_x = np.log(data[1923, 1])
min_temp_star_y = np.log(data[1923, 0])
plt.scatter(min_temp_star_x, min_temp_star_y, color = "orange")

plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")

plt.show
```

```python
# Give scatter points colormap
plt.plot(figsize = (8, 6))
colored_points = plt.scatter(log_temperature, log_magnitude, c=temperature, cmap="viridis")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")

my_colorbar = plt.colorbar()
my_colorbar.set_label("Temperature K")
```
