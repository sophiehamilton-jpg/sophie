```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# --- 2 Matplotlib ---

# create function
def make_sin_wave(x, A, w):
    y = A * np.sin(w * x)
    return y
```

```python
# define x, A, and w
x = np.linspace(0, 2*np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
# loop print frequencies and amplitudes
for amplitude, frequency in zip(amplitude, frequency):
    print (f"Amplitude: {amplitude}, frequency: {frequency}")
```

```python
# plot graphs
fig, ax = plt.subplots(1, 3, figsize = (25, 5))

sin_wave_1 = make_sin_wave(x, amplitude, 1)
sin_wave_2 = make_sin_wave(x, 1, frequency)
sin_wave_3 = make_sin_wave(x, amplitude, frequency)


ax[0].plot(x, sin_wave_1, linestyle = "--")
ax[1].plot(x, sin_wave_2, linestyle = ":")
ax[2].plot(x, sin_wave_3, linestyle = "solid")

# label axes and title
ax[0].set_title("Sin Curve: Varying Amplitude", fontsize = 20)
ax[0].set_xlabel("X", fontsize = 20)
ax[0].set_ylabel("Y", fontsize = 20)

ax[1].set_title("Sin Curve: Varying Frequency", fontsize = 20)
ax[1].set_xlabel("X", fontsize = 20)
ax[1].set_ylabel("Y", fontsize = 20)

ax[2].set_title("Sin Curve: Varying Amplitude and Frequency", fontsize = 20)
ax[2].set_xlabel("X", fontsize = 20)
ax[2].set_ylabel("Y", fontsize = 20)


# show graphs
plt.show()
```
