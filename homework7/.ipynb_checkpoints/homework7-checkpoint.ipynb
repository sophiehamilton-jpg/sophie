```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# --- 2 Matplotlib ---

# create function
def make_sin_wave(x, A, w):
    y = A * np.sin(w * x)
    return y
```

```python
# define x, A, and w
x = np.linspace(0, 2*np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

```

```python
# print frequencies and amplitudes
for amplitude, frequency in zip(amplitude, frequency):
    print (f"Amplitude: {amplitude}, frequency: {frequency}")
```

```python
# plot graphs
fig, ax = plt.subplots(1, 3, figsize = (25, 5))

sin_wave_1 = make_sin_wave(x, amplitude, 1)
sin_wave_2 = make_sin_wave(x, 1, frequency)
sin_wave_3 = make_sin_wave(x, amplitude, frequency)


ax[0].plot(x, sin_wave_1)
ax[1].plot(x, sin_wave_2)
ax[2].plot(x, sin_wave_3)

# label axes
x_label = "x"
y_label = "y"

# show graphs
plt.show()
```
