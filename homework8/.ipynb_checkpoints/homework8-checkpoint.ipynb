```python
import pandas as pd
```

```python
df = pd.read_csv("state_data.csv")
```

```python
print(df.tail())
```

```python
#   1.  The first column gives the two-letter state abbreviation, the next two columns give the state's 
# latitude and longitude, and the last column gives the full name of the state
#   2.  The first state is Alaska
#   3.  The last state is Wyoming
#   4. There are 52 total entries
#   5.  The list also contains Puerto Rico and the District of Columbia
```

```python
df["latitude"].median()
# 39.4355155 is the median latitudew
```

```python
df["longitude"].median()
# -89.093198 is the median longitude
```

```python
df["distance to median latitude"] = (df["latitude"].median() - df["latitude"]).abs()
df
```

```python
df["distance to median longitude"] = (df["longitude"].median() - df["longitude"]).abs()
df
```

```python
print(df.shape)
# 52 rows and 6 columns
```

```python
smallest_latitude_difference = 100
for state in df["distance to median latitude"]:
    if state < smallest_latitude_difference:
        smallest_latitude_difference = state
print(f"\033[1mThe state closest to the median latitude is Utah by {smallest_latitude_difference:.2f}\033[0m")
```

```python
greatest_latitude_difference = 0
for state in df["distance to median latitude"]:
    if state > greatest_latitude_difference:
        greatest_latitude_difference = state
print(f"\033[1mThe state farthest from the median latitude is Alaska by {greatest_latitude_difference:.2f}\033[0m")
```

```python
smallest_longitude_difference = 100
for state in df["distance to median longitude"]:
    if state < smallest_longitude_difference:
        smallest_longitude_difference = state
print(f"\033[1mThe state closest from the median latitude is Mississippi by {smallest_longitude_difference:.2f}\033[0m")
```

```python
greatest_longitude_difference = 0
for state in df["distance to median longitude"]:
    if state > greatest_longitude_difference:
        greatest_longitude_difference = state
greatest_longitude_difference
print(f"\033[1mThe state farthest from the median longitude is Hawaii by {greatest_longitude_difference:.2f}\033[0m")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("titanic")
df
```

```python
df = df.dropna(axis = 1)
```

```python
df.shape
# the shape of the dataset is (891, 11), meaning it has 891 rows and 15 columns
```

```python
print(df.head())
```

```python
print(df.tail())
```

```python
# --- Column Names
# 1. survived: value of 1 if passenger survived sinking, value of 0 if passenger did not survive
# 2. pclass: ticket class (1st-3rd)
# 3. sex: passenger's sex (male or female)
# 4. sibsp: number of siblings/spouses onboard
# 5. parch: number of parents/children onboard
# 6. fare: cost of passenger's ticket in USD
# 7. class: class on board as string (First-Third)
# 8. who: gender (man or woman)
# 9. adult_male: True if passenger was an adult male, otherwise False
# 10. alive: "yes" if passenger survived sinking, "no" if passenger did not survive
# 11. alone: True if no one was accompanying them, otherwise False 
```

```python
fig, ax = plt.subplots(2, 2, figsize = (12, 12))

x1 = df["pclass"]
y1 = df["fare"]

x2 = df["alive"]
y2 = df["fare"]

ax[0, 0].scatter(x1, y1)
ax[0, 0].set_title("Class vs Ticket Cost")
ax[0, 0].set_ylabel("Price of Ticket (USD)")
ax[0, 0].set_xlabel("Class")

ax[0,1].scatter(x2, y2, c = "g")
ax[0, 1].set_title("Survival vs Ticket Cost")
ax[0, 1].set_ylabel("Price of Ticket (USD)")
ax[0, 1].set_xlabel("Alive")

sns.histplot(df["alive"], ax = ax[1,0], color = "red")
ax[1, 0].set_title("Count of Passengers Alive and Dead")
ax[1, 0].set_ylabel("Number of Passengers")
ax[1, 0].set_xlabel("Alive")


sns.boxplot(df["sibsp"], ax = ax[1, 1], color = "orange")
ax[1, 1].set_title("Numbers of Siblings/Spouces per Passenger")
ax[1, 1].set_ylabel("Number of Simplers or Spouces")
ax[1, 1].set_xlabel("")
```

```python
# Explaining the graphs:

# 1. The first plot shows the price of ticket cost as a function of passenger's class. 
#    I chose these columns because I wanted to see if there was a dramatic difference in
#    ticket price between the classes. I was surprised to see that the distribution in 
#    price was similar between 2nd and 3rd class passengers.
# 2. The second plot shows the price of tickets of surviving passengers vs the price of 
#    tickets of dead passengers. I wanted to see if price of ticket had a signifigant impact
#    on chance of survival but the data indicates that passengers of all classes died.
# 3. The third plot shows how many passengers survived. I learned that more than half of
#    the passengers onboard died.
# 4. The forth plot shows how many siblings and spouces passengers had on the ship. I was
#    curious about what the distribution would look like, and the boxplot indicates that 
#    most people had 0 - 1 siblings or spouces onboard. One person had 8!
```
