```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
df = pd.read_csv("solar_system.csv")
```

```python
df.shape
# Before transposing, the DataFrame has 20 rows and 11 columns
```

```python
df = df.set_index("Attribute").T # makes the values in the column "attribute" into row labels, swaps rows and columns
df.index.name = "Planet" # names the row labels "planet"
df.reset_index(inplace = True) # converts the row labels into a column
df.columns.name = None # removes the column's name
```

```python
df.shape
# After transposing, the DataFrame has 10 rows and 21 columns
# The shape changed because the axes were flipped and the column label had to be changed
```

```python
print(df.columns)
# 16 columns have units: Mass (10^24kg), Diameter (km), Density (kg/m^3), Gravity (m/s^2), Escape Velocity (km/s), 
#     Rotation Period (hours), Length of Day (hours), Distance from Sun (10^6 km), Perihelion (10^6 km), 
#     Aphelion (10^6 km), Orbital Period (days), Orbital Velocity (km/s), Orbital Inclination (deg), 
#     Obliquity to Orbit (deg), Mean Temperature (C), Surface Pressure (bars)
# 5 columns don't have units: Planet, Orbital Eccentricity, Number of Moons, Ring System?, Global Magnetic Field? 
```

```python
for col in df.iloc[:, 1:20]:
    df[col] = pd.to_numeric(df[col], errors = "coerce")
```

```python
df.dtypes
# (a) The three different data types now in the DataFrame are string, float, and integer
# (b) A string is a non-numberical collection of symbols, a float is a number with decimal places, and an integer is a whole number
# (c) The planet column is still object because its string was not numerical
```

```python
def attach_units(df, old_col, unit, new_col = None):
    df[old_col] = [val * unit for val in df[old_col]]
    if new_col is not None:
        df.rename(columns={old_col: new_col}, inplace = True)
    return df
```

```python
my_unit = 1e6 * u.km
my_old_col = "Distance from Sun (10^6 km)"
my_new_col = "Distance from Sun (km)"

attach_units(df, my_old_col, my_unit, my_new_col)
```
