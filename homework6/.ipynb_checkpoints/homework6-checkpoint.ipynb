```python
# 4: Prime star systems
```

```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8], 
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime(num):
    if num == 2:
        return True
    for i in range(2, num):
        if num % i == 0:
            return False
        else:
            return True

def prime_rows(my_array):
    new_array = []
    for i in my_array:
        if any(is_prime(k) for k in i):
                new_array.append(i)
    return np.array(new_array)

print(prime_rows(arr))
```

```python
# 5: Let's play checkers
```

```python
# 5.1
def blank_square(n):
    return np.empty((n, n), dtype = int)

print(blank_square(8))
```

```python
# 5.2
def half_checkerboard(n):
    board = np.empty((n, n), dtype = int)
    pattern = ([1, 0] * 4)
    board[::2] = pattern
    return board
print(half_checkerboard(8))
```

```python
# 5.3
def checkerboard(n):
    board = np.empty((n, n), dtype = int)
    pattern1 = ([1, 0] * 4)
    pattern2 = ([0, 1] * 4)
    board[::2] = pattern1
    board[1::2] = pattern2
    return board

print(checkerboard(8))
```

```python
# 5.4
def reverse_checkerboard(n):
    board = np.empty((n, n), dtype = int)
    pattern1 = ([0, 1] * 4)
    pattern2 = ([1, 0] * 4)
    board[::2] = pattern1
    board[1::2] = pattern2
    return board

print(reverse_checkerboard(8))
```

```python
# 6 The expanding universe
```

```python
def expansion(arr, num):
    spaces = " " * num
    return np.array([spaces.join(word) for word in arr])
    
universe = np.array(["galaxy", "clusters"])

print(expansion(universe, 1))
print(expansion(universe, 3))
```

```python
# 7 Second-Brightest Star
```

```python
np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
print(stars)
```

```python
def secondBrightest(arr):
    new_array = []
    for row in arr:
        min_mag = np.min(row)
        second_lowest_mag = 2000
        for star in row:
            if star > min_mag and star < second_lowest_mag:
                second_lowest_mag = star
        new_array.append(second_lowest_mag)
    return np.array(new_array)

print(secondBrightest(stars))
```
