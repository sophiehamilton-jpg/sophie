```python
import skyfield
from skyfield.api import Star, load
from skyfield.data import hipparcos, stellarium
from skyfield.projections import build_stereographic_projection
from astropy.coordinates import SkyCoord
import astropy.units as u
from astropy.wcs import WCS
from astropy.visualization.wcsaxes import WCSAxes
```

```python
with load.open(hipparcos.URL) as f:
    df = hipparcos.load_dataframe(f)
```

```python
df = df[df["magnitude"] <= 2.5]
```

```python
# define earth and planets
planets = load("de421.bsp")
earth = planets["earth"]
```

```python
# define timescale
ts = load.timescale()
t = ts.utc(2018, 9, 3)

bright_stars = Star.from_dataframe(df)
astrometric = earth.at(t).observe(bright_stars)
ra, dec, distance = astrometric.radec() # ra, dec denote coordinate position
```

```python
fig, ax = plt.subplots()

ax.scatter(ra.hours, dec.degrees, 8 - df['magnitude'], "w")
ax.set_xlim(7.0, 4.0)
ax.set_ylim(-20, 20)
ax.grid(True)
ax.set(title='The brightest stars in Orion',
       xlabel='Right ascension (h)', ylabel='Declination (°)')

plt.style.use("dark_background")
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman']

plt.savefig('bright_stars.png', dpi=100, facecolor='#1a1a1a')
```
