```python
# Matplotlib for Graphing
import matplotlib.pyplot as plt

#Skyfield
from skyfield.api import Star, load, wgs84
from skyfield.data import hipparcos
from skyfield.projections import build_stereographic_projection

# Location and time libraries
import datetime as dt
from datetime import datetime # allows us to manipulate dates and times
from tzwhere import tzwhere # gives timezone for a latitude and longitude
import pytz
from pytz import timezone, utc # timezone data
from geopy.geocoders import Nominatim # can convert place names into geographic coordinates
```

```python
# Load in the hipparcos star data
with load.open(hipparcos.URL) as f:
    df = hipparcos.load_dataframe(f)

# define earth and planets
planets = load("de421.bsp")
earth = planets["earth"]
```

```python
# --- Variables ---

# Define time and location on Earth
time_str = "2000-01-01 00:00"
location_name = "Claremont, CA"

# --- Finding Latitude and Longitude ---

# Access the latitude and longitude of the location
geolocator = Nominatim(user_agent="my_geocoding_app")
location = geolocator.geocode(location_name)
lat, long = location.latitude, location.longitude

# Print the latitude and longitude
if location:
    print(f"Address: {location.address}")
    print(f"Latitude: {location.latitude}")
    print(f"Longitude: {location.longitude}")
else:
    print(f"Could not find coordinates for '{location_name}'")
```

```python
# --- Access the timezone of our location ---

timezone_str = tzwhere.tzwhere().tzNameAt(lat, long)
print(f"Timezone Name: {timezone_str}")

dt_naive = datetime.strptime(time_str, '%Y-%m-%d %H:%M') # converts our time value into an objet that python can read

# converts the timezone of our location to universal timezone
local_tz = pytz.timezone(timezone_str)
local_dt_aware = local_tz.localize(dt_naive, is_dst=None)
print(f"\nLocal Aware Datetime: {local_dt_aware}")
utc_dt = local_dt_aware.astimezone(pytz.utc)
print(f"UTC Datetime: {utc_dt}")
```
